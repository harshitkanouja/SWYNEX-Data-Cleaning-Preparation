```python
monthly_cols = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN',
                'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']

print(df[monthly_cols].isnull().any(axis=1).sum())
```

```python
df_cleaned= df.dropna(subset=monthly_cols).copy()
```

```python
print("Original rows:", len(df))
print("Cleaned rows:", len(df_cleaned))
print("Rows removed:", len(df) - len(df_cleaned))
```

```python
print(df_cleaned[monthly_cols].isnull().sum())
```

```python
print("Total missing values:", df_cleaned.isnull().sum().sum())
```

```python
print(df_cleaned.dtypes)
```

```python
monthly_total = df_cleaned[monthly_cols].sum(axis=1)

difference = abs(df_cleaned["ANNUAL"] - monthly_total)

print("Maximum difference:", difference.max())
```

```python
print("Rows with difference:", (difference > 0.1).sum())
```

```python
print(difference.describe())
```

```python
print(difference.sort_values(ascending=False).head(10))
```

```python
df_cleaned["JF_calc"] = df_cleaned["JAN"] + df_cleaned["FEB"]

df_cleaned["MAM_calc"] = (
    df_cleaned["MAR"] +
    df_cleaned["APR"] +
    df_cleaned["MAY"]
)

df_cleaned["JJAS_calc"] = (
    df_cleaned["JUN"] +
    df_cleaned["JUL"] +
    df_cleaned["AUG"] +
    df_cleaned["SEP"]
)

df_cleaned["OND_calc"] = (
    df_cleaned["OCT"] +
    df_cleaned["NOV"] +
    df_cleaned["DEC"]
)
```

```python
print("JF max difference:",
      abs(df_cleaned["JF"] - df_cleaned["JF_calc"]).max())

print("MAM max difference:",
      abs(df_cleaned["MAM"] - df_cleaned["MAM_calc"]).max())

print("JJAS max difference:",
      abs(df_cleaned["JJAS"] - df_cleaned["JJAS_calc"]).max())

print("OND max difference:",
      abs(df_cleaned["OND"] - df_cleaned["OND_calc"]).max())
```

```python
df_cleaned.drop(
    columns=["JF_calc", "MAM_calc", "JJAS_calc", "OND_calc"],
    inplace=True
)
```

```python
print(df_cleaned.columns)
```

```python
df_cleaned.describe()
```

```python
print("Total Subdivisions:", df_cleaned["SUBDIVISION"].nunique())

print("\nSubdivisions:")
print(df_cleaned["SUBDIVISION"].unique())
```

```python
print("Starting Year:", df_cleaned["YEAR"].min())
print("Ending Year:", df_cleaned["YEAR"].max())
```

```python
yearly_rainfall = (
    df_cleaned.groupby("YEAR")["ANNUAL"]
    .mean()
    .reset_index()
)

print(yearly_rainfall.head())
print(yearly_rainfall.tail())
```

```python
print("Number of years:", yearly_rainfall["YEAR"].nunique())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(
    yearly_rainfall["YEAR"],
    yearly_rainfall["ANNUAL"]
)

plt.xlabel("Year")
plt.ylabel("Average Annual Rainfall (mm)")
plt.title("Year-wise Average Annual Rainfall (1901–2017)")

plt.grid(True, alpha=0.3)
plt.show()
```

```python
highest_years = yearly_rainfall.sort_values(
    by="ANNUAL",
    ascending=False
).head(10)

lowest_years = yearly_rainfall.sort_values(
    by="ANNUAL"
).head(10)

print("Top 10 Highest Rainfall Years")
print(highest_years)

print("\nTop 10 Lowest Rainfall Years")
print(lowest_years)
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    highest_years["YEAR"].astype(str),
    highest_years["ANNUAL"]
)

plt.xlabel("Year")
plt.ylabel("Average Rainfall (mm)")
plt.title("Top 10 Highest Rainfall Years")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    lowest_years["YEAR"].astype(str),
    lowest_years["ANNUAL"]
)

plt.xlabel("Year")
plt.ylabel("Average Rainfall (mm)")
plt.title("Top 10 Lowest Rainfall Years")

plt.xticks(rotation=45)
plt.show()
```

```python
Q1 = df_cleaned["ANNUAL"].quantile(0.25)
Q3 = df_cleaned["ANNUAL"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df_cleaned[
    (df_cleaned["ANNUAL"] < lower_bound) |
    (df_cleaned["ANNUAL"] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

```python
print(
    outliers[["SUBDIVISION", "YEAR", "ANNUAL"]]
    .sort_values("ANNUAL", ascending=False)
    .head(10)
)
```
